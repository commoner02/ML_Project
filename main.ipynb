```python
# --- Column 18: Yes/No ---
yes_no_col = '18. Do you usually have a second sleep period during the day or at another time?\n(আপনি কি সাধারণত দিনে বা অন্য কোনো সময় দ্বিতীয়বার ঘুমান?)'
```

```python

# Step 1: Check unique values
print("Unique values in Second Sleep Yes/No column:")
print(df[yes_no_col].unique())
```

```python

# Step 2: Map to binary
df['Has_Second_Sleep'] = df[yes_no_col].map({'Yes': 1, 'No': 0})
```

```python

# Step 3: Check for unmapped values
unmapped = df[df['Has_Second_Sleep'].isna()]
if len(unmapped) > 0:
    print(f"\nWarning: {len(unmapped)} rows had unmapped values.")
else:
    print("\nYes/No mapped successfully!")
```

```python

# --- Column 19: Second Sleep Start Time ---
start_col = '19. If Yes, What time does your second sleep usually start?(হ্যাঁ হলে, আপনার দ্বিতীয়বার ঘুমানো সাধারণত কখন শুরু হয়?'
```

```python

# Step 4: Convert time strings to numeric hours
def time_to_hours(t_str):
    if pd.isna(t_str):
        return np.nan
    # Parse "4:00:00 PM" format
    dt = pd.to_datetime(t_str, format='%I:%M:%S %p', errors='coerce')
    if pd.isna(dt):
        return np.nan
    return dt.hour + dt.minute / 60.0

df['SecondSleep_Start_Hour'] = df[start_col].apply(time_to_hours)
```

```python

# --- Column 20: Second Sleep End Time ---
end_col = '20. If Yes, What time do you usually wake up from your second sleep?(হ্যাঁ হলে, দ্বিতীয়বার ঘুমানোর পর আপনি সাধারণত কখন ঘুম থেকে ওঠেন?)'
```

```python

df['SecondSleep_End_Hour'] = df[end_col].apply(time_to_hours)
```

```python

# Step 5: Show comparison
print("\nFirst 10 rows (original -> new):")
print(df[[yes_no_col, start_col, end_col, 'Has_Second_Sleep', 'SecondSleep_Start_Hour', 'SecondSleep_End_Hour']].head(10))
```

```python
# Cell: One-Hot Encode Department, Academic Year, and Device

# List of columns to encode
cat_cols = [
    '1. Department (বিভাগ)',
    '2. Which academic year are you currently in?(আপনি বর্তমানে কোন শিক্ষাবর্ষে অধ্যয়ন করছেন?)',
    '3. Which digital device do you use most frequently?(আপনি কোন ডিজিটাল ডিভাইসটি সবচেয়ে বেশি ব্যবহার করেন?)'
]
```

```python

# Step 1: Check unique counts
print("Unique counts before encoding:")
for col in cat_cols:
    print(f"  {col}: {df[col].nunique()} unique values")
```

```python

# Step 2: Apply one-hot encoding
# drop_first=False keeps all categories (so we know exactly what each column represents)
df = pd.get_dummies(df, columns=cat_cols, prefix=['Dept', 'Year', 'Device'], drop_first=False)
```
